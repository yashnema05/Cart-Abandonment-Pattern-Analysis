# Cart Abandonment Pattern Analysis & AI-Based Prediction

Core Python workflow for the project.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
```

```python
df = pd.read_csv('/content/Ecommerce.csv')
print('Rows:', df.shape[0])
print('Columns:', df.shape[1])
print(df.info())
```

```python
print(f"Abandonment Rate: {df['cart_abandoned'].mean():.2%}")
print('Total Sessions:', len(df))
print('Added to Cart:', int(df['added_to_cart'].sum()))
print('Purchases:', int(df['purchased'].sum()))
```

```python
duration_mapping = {'Very Short': 0, 'Short': 1, 'Long': 2, 'Very Long': 3}
df['session_duration_encoded'] = df['session_duration_bucket'].map(duration_mapping)
features = ['device_type','user_type','marketing_channel','product_category','unit_price','quantity','discount_percent','discount_amount','pages_viewed','time_on_site_sec','rating','payment_method','visit_weekday','visit_season','session_duration_encoded']
X = df[features]
y = df['cart_abandoned']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)
```

```python
model = RandomForestClassifier(n_estimators=200, random_state=42, class_weight='balanced')
model.fit(X_train, y_train)
pred = model.predict(X_test)
print('Accuracy:', accuracy_score(y_test, pred))
print(classification_report(y_test, pred))
```

```python
importance = pd.DataFrame({'Feature': features, 'Importance': model.feature_importances_}).sort_values('Importance', ascending=False)
print(importance)
```

## Final Results

- Abandonment rate: **42.00%**
- Added to cart: **16,117 / 25,000**
- Purchases: **5,616 / 25,000**
- Random Forest accuracy: **58.32%**
- Top features: **unit price, time on site, rating, pages viewed, discount amount**
